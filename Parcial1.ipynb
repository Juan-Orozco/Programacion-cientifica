### Ejercicio #6

```python
lecturas_pm25 = [8.4, 12.7, -999.0, 15.2, 41.8, 9.6, -999.0, 58.3, 22.0, 13.9, 7.5, 36.1]
lecturas_validas = []
suma_pm25 = 0.0
lecturas_descartadas  = 0



for lectura in lecturas_pm25:
    if lectura == -999.0:
        lecturas_descartadas +=1
        continue

    lecturas_validas.append(lectura)
    suma_pm25 += lectura

promedio_pm25 = suma_pm25  / len(lecturas_validas)

print(lecturas_validas)
print(lecturas_descartadas)
print(promedio_pm25)
```

### Ejercicio #7

```python
maximo_pm25 = lecturas_validas[0]
minimo_pm25 = lecturas_validas[0]

for lectura in lecturas_validas:
    if lectura > maximo_pm25:
        maximo_pm25 = lectura
    if lectura < minimo_pm25:
        minimo_pm25 = lectura

rango = maximo_pm25 - minimo_pm25

suma_desviaciones = 0.0

for lectura in lecturas_validas:
    diferencia = lectura - promedio_pm25
    suma_desviaciones += diferencia ** 2

cociente = suma_desviaciones / len(lecturas_validas)
desviacion_estandar = cociente ** 0.5


print("Máximo: "+ str(maximo_pm25))
print("Mínimo: " + str(minimo_pm25))
print("Rango:"+ str(rango))
print("Desviación Estándar: " + str(desviacion_estandar))
```
